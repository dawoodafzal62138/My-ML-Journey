# **Pandas**
**Pandas** (Panel data) is a popular Python library used for **data manipulation and analysis**. It makes it easy to work with structured data such as spreadsheets, CSV files, SQL tables, and JSON.

The two main data structures in Pandas are:

* **Series**: A one-dimensional labeled array (similar to a single column in a spreadsheet).
* **DataFrame**: A two-dimensional table with rows and columns (similar to an Excel worksheet or SQL table).

### What can you do with Pandas?

* Read data from files (CSV, Excel, JSON, SQL, etc.)
* Clean and preprocess data (handle missing values, remove duplicates)
* Filter, sort, and group data
* Merge and join datasets
* Perform statistical analysis
* Create summaries and reports
* Prepare data for machine learning

##  Series
-  A series is a one-dimensional labeled array capable of holding any data type (integers, strings, floating point numbers, Python objects, etc.). It can be thought of as a single column in a spreadsheet or a database table.


```python
data = [101,102,103,104 ,105]
series  = pd.Series(data) # -> Series() is a constructor
# print(series)

# we can also change the label using index keyword in constructor

series_1 = pd.Series(data, index = ["room # 1" , "room # 2" ,"room # 3" ,"room # 4" ,"room # 5" ] )
print(series)
print()

# we can access the values of the series using the index labels
print(f"The value of room # 1 is {series_1['room # 1']}")
print(f"The value of room # 2 is {series_1['room # 2']}")

# or multiple values can be accessed using a list of index labels
print(f"The value of room # 1 and room # 2 is \n{series_1[['room # 1' , 'room # 2']]}")




```

```python
# using numpy like funtions on series 
series_2  =pd.Series(np.arange(1,11), index = ["a", "b", "c", "d", "e", "f", "g", "h", "i", "j"])
print(f"Series 2 is \n{series_2}")

print(f"\nValues greater than 5 \n{series_2[series_2 > 5]}")  
print(f"\nValues squared \n{series_2**2}")  

print(f"if a in series_2: {'a' in series_2}")
print(f"if k in series_2: {'k' in series_2}")
```

```python

# array representation of series can be obtained using the array attribute
print(f"\nThe array representation of the series is {series_1.array}")

# the index of the series can be obtained using the index attribute
print(f"\nThe index of the series is {series_1.index}")



# series can also be converted to a numpy array using the to_numpy() method
series_array = series_1.to_numpy()
print(f"\nThe series converted to a numpy array is {series_array}\n")

```

## Reindexing

```python
# an important fearture of pandas is that it pandas object can reindex , which means to create the new values rearranged to align with the new indexes
series = pd.Series([1,2,3,4,5], index = ["a", "b", "c", "d", "e"])
print(f"\nOriginal series : \n{series}")



 # if the new index is not present in the original series then it will be filled with NaN values
print(f"\nReindexed series : \n{series.reindex(['b' , 'c' , 'a' , 'd' , 'e' , 'f'])}")
```

```python
# ffil method can be used to fill the values when reindexing
colors = pd.Series(["red" , "blue" , "green"], index = [0,2,4])
print(f"\nOriginal colors series : \n{colors}")




print(f"\nReindexed colors series with ffill method : \n{colors.reindex(range(6), method = "ffill")}")


```

```python
# similarly in dataframe reindex can be done which alter row , columns or both 

student_data = {
    "name" : ["Ali" , "Ahmed" , "Ayesha" , "Zainab" , "Hassan"],
    "marks" : [50,60,65,94,78],
    "age" : [20,21,22,23,24],

}
student_df = pd.DataFrame(student_data , index = ["a" , "b" , "c" , "d" , "e"])
print(f"\nOriginal student dataframe : \n{student_df}")


reindexed_student_df = student_df.reindex(labels=["b" , "c" , "a" , "d" , "e" , "f"] , columns=["gender" , "marks" , "age" , "name"])

print(f"\nReindexed student dataframe : \n{reindexed_student_df}")
```

**Some arguments of .reindex()**
| Argument     | Description                                                                                                                    |
| ------------ | ------------------------------------------------------------------------------------------------------------------------------ |
| `labels`     | New labels for the rows or columns. What it affects depends on the `axis` argument.                                            |
| `index`      | Specifies the new row labels/index.                                                                                            |
| `columns`    | Specifies the new column labels.                                                                                               |
| `axis`       | Specifies whether `labels` refers to the **index (rows)** or **columns**. `0` / `"index"` = rows, `1` / `"columns"` = columns. |
| `method`     | Determines how missing values are filled when reindexing. Common options: `"ffill"`, `"bfill"`, `"nearest"`.                   |
| `copy`       | Controls whether a copy of the data should be made.                                                                            |
| `level`      | Used with a **MultiIndex** to specify which level should be reindexed.                                                         |
| `fill_value` | Value used to fill newly introduced missing positions. Default is `NaN`.                                                       |
| `limit`      | Maximum number of consecutive missing values to fill when using `method`.                                                      |
| `tolerance`  | Maximum distance between the existing index and the new index when using methods such as `"nearest"`.                          |
