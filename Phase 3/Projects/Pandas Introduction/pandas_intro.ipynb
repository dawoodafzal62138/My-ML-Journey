# **Pandas**
**Pandas** (Panel data) is a popular Python library used for **data manipulation and analysis**. It makes it easy to work with structured data such as spreadsheets, CSV files, SQL tables, and JSON.

The two main data structures in Pandas are:

* **Series**: A one-dimensional labeled array (similar to a single column in a spreadsheet).
* **DataFrame**: A two-dimensional table with rows and columns (similar to an Excel worksheet or SQL table).

### What can you do with Pandas?

* Read data from files (CSV, Excel, JSON, SQL, etc.)
* Clean and preprocess data (handle missing values, remove duplicates)
* Filter, sort, and group data
* Merge and join datasets
* Perform statistical analysis
* Create summaries and reports
* Prepare data for machine learning



```python
import pandas as pd
import pathlib
```

## 1. Series

```python
data = [101,102,103,104 ,105]
series  = pd.Series(data) # -> Series() is a constructor
# print(series)

# we can also change the label using index keyword in constructor

series_1 = pd.Series(data, index = ["room # 1" , "room # 2" ,"room # 3" ,"room # 4" ,"room # 5" ] )
# print(series)

# we can also create series from a dictionary
pizza_flavours = {
    "flavour 1" : "margrita",
    "flavour 2" : "pepperoni",
    "flavour 3" : "Cheese", 
    "flavour 4" : "meat lover"
}


series_2 = pd.Series(pizza_flavours)

print(series_2)

# we can also access value by label and indexes by using loc (for labels) and iloc (for index)

print(f"\nMy favourite pizza flavour is {series_2.loc["flavour 2"]}")

print(f"\nMy friends favourite pizza flavour is {series_2.iloc[0]}")



```

```python
# changing the value 
series_2.loc["flavour 4"] = "B.B.Q"
print(f"changed flavour : {series_2.loc["flavour 4"]}")
```

```python
# Filtering the data


students_marks = [50,60,65,94,78,83,42]
series_3 = pd.Series(students_marks)

print(f"Students marks less than 60 : \n{series_3[series_3 < 60]}")
```

## 2. Data Frame

```python
employee = {
    "names" : ["spongebob" , "patrik" , "squidward"],
    "age" : [30,35,50],
    "job" : ["cook" , "gardener" , "manager"]
}


df = pd.DataFrame(employee , index=["employee 1" ,"employee 2" ,"employee 3" ,])

print(df)

# Adding a new column 
df["salary"] = [20000 , 10000 , 40000]

# print(df)

print(f"\n\nEmployee 1 detail : \n{df.loc["employee 1"]}")


print(f"\n\nEmployee whose salary is greater than 15000 \n{df[df['salary'] > 15000]}")



# adding a new row in a data frame

new_employees = pd.DataFrame([{"names" : "sandy" ,  "age" : 28 , "job" : "scientist" , "salary" : 80000 },
                             {"names" : "Mr. crabs" ,  "age" : 60 , "job" : "owner" , "salary" : 100000 }]
                             , index= ["employee 4" , "employee 5"])

df = pd.concat([df , new_employees ])

print(f"\n\nAfter adding new employees : \n{df}")

```

# 7. Aggretation
* Reduces a set of values into a single summary value
* Used to summarize and analyze data
* Often used woth the groupby( ) function

```python
# For whole dataframe
print(f"Mean of all numeric Columns: \n{df.mean(numeric_only= True)}")
print(f"\nMinimum of all numeric Columns: \n{df.min(numeric_only= True)}")
print(f"\nMaximum of all numeric Columns: \n{df.max(numeric_only= True)}")
print(f"\nCount of all numeric Columns: \n{df.count(numeric_only= True)}")
print(f"\nSum of all numeric Columns: \n{df.sum(numeric_only= True)}")
```

```python
# Single Column 
print(f"Mean of Height Column: \n{df["Height"].mean()}")
print(f"\nMinimum of Height Column: \n{df["Height"].min()}")
print(f"\nMaximum of Height Column: \n{df["Height"].max()}")
print(f"\nCount of Height Column: \n{df["Height"].count()}")
print(f"\nSum of Height Column: \n{df["Height"].sum()}")
```

```python
# Group by ( ) function
group = df.groupby("Type1")
print(f"Group by heights")
print(f"\nsum :\n{group["Height"].sum()}")
print(f"\nmin :\n{group["Height"].min()}")
print(f"\nmax :\n{group["Height"].max()}")
print(f"\nmean :\n{group["Height"].mean()}")
```

# 8. Data Cleaning
* The process of fixing/removing any incomplete , incorrect or irrelevant data

```python
df =  pd.read_csv(r"Phase 3\Projects\Pandas Introduction\data.csv")


# 1. Drop irrelevant columns



df = df.drop(columns=["No" , "Legendary"])
print(f"Dropping irrelevant columns : \n{df}")
```

```python
# 2. Handling missing data
# df = df.dropna(subset=["Type2"])

# changing a nan value with other Value
df =  df.fillna({"Type2" : "None"})
# print(df.to_string())


```

```python
# 3. Fix incosistent value
df["Type1"]  = df["Type1"].replace({"Grass" : "GRASS" , 
                                                 "Water" : "WATER"})

print(df.to_string())
```

```python
# 4. Standardiza text
df["Name"] = df["Name"].str.upper()
print(df)
```
