# **linear Regression from scratch**


1. Import the Essential libraries

```python
import numpy as np
import pandas as pd
import pathlib
from numpy.typing import NDArray


```

```python

w = 0
b = 0
data = pd.read_csv("data.csv")

x_raw = data["house_size"].values
Y_raw = data["price"].values

def min_max_normalization(array : NDArray) -> NDArray:
        return (array - np.min(array)) / (np.max(array) - np.min(array))

x = min_max_normalization(np.array(x_raw))
Y = min_max_normalization(np.array(Y_raw))

costs = 0

while True:
    y = w * x + b

    error = (1 / (2 * 100)) * (np.sum((y - Y) ** 2))

    j_w = (1 / 100) * (np.sum((y - Y) * x))
    j_b = (1 / 100) * (np.sum(y - Y))

    w -= 0.1 * j_w
    b -= 0.1 * j_b
    if costs == error:
          break
    costs = error
print(f"w     = {w:.4f}")
print(f"b     = {b:.4f}")
print(f"cost  = {costs}")

```
