# **linear Regression from scratch**


### 1. Import the Essential libraries

```python
import numpy as np
import pandas as pd
import pathlib
from numpy.typing import NDArray


```

### 2. Read the data from csv file and seperate the values(house_prize , house_size)


```python
data = pd.read_csv("data.csv")
x_raw = data["house_size"].values[:80]
Y_raw = data["price"].values[:80]
print(x_raw.size)
```

### 3. Set the weight (w) and bias(b) to 0 initailly


```python
w = 0
b = 0
```

### 4. Define a normalize function that normalize the data because it is too large

```python
def min_max_normalization(array : NDArray) -> NDArray:
        return (array - np.min(array)) / (np.max(array) - np.min(array))
def min_max_denormalization(array : NDArray , array_norm: NDArray) -> NDArray:
        return array_norm * (np.max(array) - np.min(array)) + np.min(array)
        # Y_predict_norm * (Y_raw.max() - Y_raw.min()) + Y_raw.min()
x = min_max_normalization(np.array(x_raw))
Y = min_max_normalization(np.array(Y_raw))
```

### 4. Actuall implementation
1. predict the values using initiall weights and bias
2. calculate error and gradients
3. Update the weights and bias


```python
costs = 0
m = 80
while True:
    y = w * x + b

    error = (1 / (2 * m)) * (np.sum((y - Y) ** 2))

    j_w = (1 / m ) * (np.sum((y - Y) * x))
    j_b = (1 / m) * (np.sum(y - Y))

    w -= 0.1 * j_w
    b -= 0.1 * j_b
    if abs(costs - error) < 1e-10:
        break
    costs = error
print(f"w     = {w:.4f}")
print(f"b     = {b:.4f}")
print(f"cost  = {costs}")
# w     = 0.8320
# b     = 0.0864
# cost  = 0.0011882965167950826
```

```python
x_test = data["house_size"].values[81:]
Y_test = data["price"].values[81:]
# normalize input
x_test_norm = min_max_normalization(np.array(x_test))

# predict
Y_predict_norm = w * x_test_norm + b

# denormalize output back to real prices
Y_predict = min_max_denormalization(Y_raw, Y_predict_norm)

for Y_t, Y_p in zip(Y_test, Y_predict):
    print(f"Actual : ${Y_t:,.0f}  |  Predicted : ${Y_p:,.0f}")
    
mae = np.mean(np.abs(Y_test - Y_predict))
print(f"MAE = ${mae:,.0f}")
```
