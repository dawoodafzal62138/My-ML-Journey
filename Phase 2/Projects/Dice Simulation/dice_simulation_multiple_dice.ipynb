##  Multiple Dice Simulation (Rolling Two Dice)

Unlike a single die where every outcome (1-6) has an equal probability ($\frac{1}{6}$), the sum of two dice has a non-uniform distribution follows a Guassian Curve. The possible outcomes range from 2 to 12.

The Theoretical Probability ($P_t$) for Two Dice:

There are $6 \times 6 = 36$ total possible combinations. The probability of each sum is calculated by dividing the number of favorable combinations by 36:

* Sum = 2 or 12: 1 combination $\rightarrow \frac{1}{36} \approx 0.0278$

* Sum = 3 or 11: 2 combinations $\rightarrow \frac{2}{36} \approx 0.0556$

* Sum = 4 or 10: 3 combinations $\rightarrow \frac{3}{36} \approx 0.0833$

* Sum = 5 or 9: 4 combinations $\rightarrow \frac{4}{36} \approx 0.1111$

* Sum = 6 or 8: 5 combinations $\rightarrow \frac{5}{36} \approx 0.1389$

* Sum = 7: 6 combinations $\rightarrow \frac{6}{36} \approx 0.1667$ (The most likely outcome!)

Let's apply the Law of Large Numbers to this scenario and simulate it to see if our experimental probability ($P_e$) converges to this bell-shaped theoretical distribution as $n \to \infty$.


## Step 2: Defining Function to Calculate probabilities:
- This function simulates random events (dice rolls) and calculates the **experimental probability** of each outcome by dividing how many times it happened by the total number of trials ($P_e = f/n$).

```python
# six sided dice 
rng= np.random.default_rng(seed=1)


def calculate_probabilities(posible_outcomes_dice1 , posible_outcomes_dice2 , number_of_rolls):


    dice1 = rng.choice(posible_outcomes_dice1 , size= number_of_rolls)
    dice2 = rng.choice(posible_outcomes_dice2 , size= number_of_rolls)
    outputs = dice1 + dice2


    element , count= np.unique(outputs , return_counts= True)

    probabilities = []

    for count_ in count:
        probability  = (count_ / number_of_rolls ) 
        probabilities.append(probability)

    return element , probabilities


```

```python
posible_outcomes_dice1 = np.arange(1, 7)
posible_outcomes_dice2 = np.arange(1, 7)

number_of_rolls = 36
element , probabilities = calculate_probabilities(posible_outcomes_dice1 , posible_outcomes_dice2, number_of_rolls)


# print(probabilities)

def draw_plot(element ,probabilities , number_of_rolls):
    plt.bar(element, probabilities, color='skyblue', edgecolor='black', width=0.6 , )

    plt.title(f'Probability Distribution of {number_of_rolls:,} Dice Rolls')
    plt.xlabel('Dice Face')
    plt.ylabel('Probability')

    plt.xticks(element)


  

    plt.show()

draw_plot(element ,probabilities ,number_of_rolls)
```

* As we can see that the probabilibilities of are not exactly like theoretical probalities
* The probalities should be equal to 1.6667

## **The Law of Large Numbers:** 
-  As we increase the number of trials, the experimental probability gets closer and closer to the theoretical probability.
$$\lim_{n \to \infty} P_e = P_t$$



* **$\lim_{n \to \infty}$** means "As the number of trials ($n$) approaches infinity..."
* **$P_e$** is your experimental probability (the messy, real-world data).
* **$P_t$** is your theoretical probability (the perfect, mathematical expectation).


```python
number_of_rolls = 100 
element , probabilities = calculate_probabilities(posible_outcomes_dice1 , posible_outcomes_dice2 , number_of_rolls)
draw_plot(element ,probabilities , number_of_rolls)
```

* As we increase the number of experiments the by 100 probability comes closer to theoritical probabilities

```python
number_of_rolls = 1000
element , probabilities = calculate_probabilities(posible_outcomes_dice1 , posible_outcomes_dice2 , number_of_rolls)
draw_plot(element ,probabilities , number_of_rolls)
```

* As we increase the number of experiments the by 1000  probability begins forming a Guassian curve. 

```python
number_of_rolls = 10000
element , probabilities = calculate_probabilities(posible_outcomes_dice1 , posible_outcomes_dice2 , number_of_rolls)
draw_plot(element ,probabilities , number_of_rolls)
```

* As we increase the number of experiments the by 10000  probability looks like forming a nearly perfect Guassian curve. 

```python
number_of_rolls = 100000
element , probabilities = calculate_probabilities(posible_outcomes_dice1 , posible_outcomes_dice2 , number_of_rolls)
draw_plot(element ,probabilities , number_of_rolls)
```

* As we increase the number of experiments the by 1000,000  probability forms a perfect Guassian curve.
