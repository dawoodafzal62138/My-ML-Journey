# **Dice Simulation**

## Step 1: Importing Libiraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

### **1. The Theoretical Probability Formula**

This is what *should* happen in a perfect world. You calculate this before you ever run an experiment. It is based on the known rules of the system (like a coin having two sides or a die having six faces).

$$P_t = \frac{\text{Number of favorable outcomes}}{\text{Total number of possible outcomes}}$$

* **$P_t$** = Theoretical probability
* **Favorable outcomes** = The specific results you are looking for.
* **Total possible outcomes** = Every possible result that could happen.

---

### **2. The Experimental Probability Formula**

This is what *actually* happens when you test it in the real world. You calculate this after you run your trials.

$$P_e = \frac{f}{n}$$

* **$P_e$** = Experimental (or empirical) probability
* **$f$** = Frequency (the number of times your desired event actually happened in your test)
* **$n$** = The total number of trials or experiments you ran


## Step 2: Defining Function to Calculate probabilities:
- This function simulates random events (dice rolls) and calculates the **experimental probability** of each outcome by dividing how many times it happened by the total number of trials ($P_e = f/n$).

```python
# six sided dice 
rng= np.random.default_rng(seed=1)


def calculate_probabilities(posible_outcomes , number_of_rolls):


    outputs = rng.choice(posible_outcomes , size= number_of_rolls)

    element , count= np.unique(outputs , return_counts= True)

    probabilities = []

    for count_ in count:
        probability  = (count_ / number_of_rolls ) 
        probabilities.append(probability)

    return element , probabilities


```

```python
posible_outcomes = np.arange(1, 7)
number_of_rolls = 10
element , probabilities = calculate_probabilities(posible_outcomes, number_of_rolls)



def draw_plot(element ,probabilities , number_of_rolls):
    plt.bar(element, probabilities, color='skyblue', edgecolor='black', width=0.6 , )

    # 2. Add labels and a title
    plt.title(f'Probability Distribution of {number_of_rolls:,} Dice Rolls')
    plt.xlabel('Dice Face')
    plt.ylabel('Probability')

    # 3. Fix the x-axis ticks to show exact dice numbers
    plt.xticks(element)


    # 4. Add a horizontal line at the theoretical probability (1/6 ≈ 0.1667)
    plt.axhline(y=1/6, color='red', linestyle='--', label='Theoretical (1/6)')
    plt.legend()

    # 5. Display the plot
    plt.show()

draw_plot(element ,probabilities ,number_of_rolls)
```

* As we can see that the probabilibilities of are 1 = 0.2, 2 = 0.2, 3 = 0.1, 4 = 0.1, 5 = 0.2, 6 = 0.2
* The probalities should be equal to 1.6667

## **The Law of Large Numbers:** 
-  As we increase the number of trials, the experimental probability gets closer and closer to the theoretical probability.
$$\lim_{n \to \infty} P_e = P_t$$



* **$\lim_{n \to \infty}$** means "As the number of trials ($n$) approaches infinity..."
* **$P_e$** is your experimental probability (the messy, real-world data).
* **$P_t$** is your theoretical probability (the perfect, mathematical expectation).


```python
number_of_rolls = 100 
element , probabilities = calculate_probabilities(element , number_of_rolls)
draw_plot(element ,probabilities , number_of_rolls)
```

* As we increase the numbers of experiment the experiment probability comes closer to theoritical probabilities

```python
number_of_rolls = 1000
element , probabilities = calculate_probabilities(element , number_of_rolls)
draw_plot(element ,probabilities , number_of_rolls)
```

* As we increase the numbers of experiment by 1000 the experiment probability comes closer to theoritical probabilities

```python
number_of_rolls = 10000
element , probabilities = calculate_probabilities(element , number_of_rolls)
draw_plot(element ,probabilities , number_of_rolls)
```

* As we increase the numbers of experiment by 10000 the experiment probability is approximately equal Theoriticalprobabilities
